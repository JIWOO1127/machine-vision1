# 파이프라인 탐색용 노트북

YOLO/OCR 라이브러리를 설치한 뒤, 이 노트북에서 이미지 한 장을 놓고
탐지 -> crop -> OCR -> 위치 매칭 각 단계를 눈으로 확인하며 디버깅할 수 있습니다.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

import cv2
import matplotlib.pyplot as plt

from classroom_locator.config import load_config
from classroom_locator.pipeline import LocatorPipeline
from classroom_locator.utils.image_utils import draw_detections
```

```python
config = load_config("../configs/default.yaml")
pipeline = LocatorPipeline(config)
```

```python
image_path = "../data/raw/sample.jpg"  # 테스트용 이미지 경로로 바꾸세요
image = cv2.imread(image_path)

result = pipeline.process_image(image)
print("탐지 개수:", len(result.detections))
print("OCR 텍스트:", result.ocr_texts_per_detection)
print("매칭 결과:", result.match)

annotated = draw_detections(image, result.detections)
plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
